```python

import pandas as pd
import numpy as np
from pathlib import Path

# Current folder 
BASE_PATH = Path(".")

# Load all CSV files
customers = pd.read_csv(BASE_PATH / "customers.csv")
accounts = pd.read_csv(BASE_PATH / "accounts.csv")
agents = pd.read_csv(BASE_PATH / "agents.csv")
transaction_types = pd.read_csv(BASE_PATH / "transaction_types.csv")
transactions = pd.read_csv(BASE_PATH / "transactions.csv")

print("All files loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Accounts": accounts,
    "Agents": agents,
    "Transaction Types": transaction_types,
    "Transactions": transactions
}

for name, df in datasets.items():
    print("\n", name)
    print("Shape:", df.shape)
    print("Duplicates:", df.duplicated().sum())
    print("Missing values:")
    print(df.isnull().sum())
```

```python
##Customers Cleaning
customers = customers.drop_duplicates()
```

```python
customers["customer_id"].duplicated().sum()
```

```python
##Remove Duplicate
customers = customers.drop_duplicates(
    subset="customer_id",
    keep="first"
)
```

```python
##Text Cleaning
customers["name"] = (
    customers["name"]
    .astype("string")
    .str.strip()
)

customers["gender"] = (
    customers["gender"]
    .astype("string")
    .str.strip()
    .str.title()
)

customers["city"] = (
    customers["city"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Date
customers["registration_date"] = pd.to_datetime(
    customers["registration_date"],
    errors="coerce"
)
```

```python
##Age Validation
customers["age"] = pd.to_numeric(
    customers["age"],
    errors="coerce"
)

customers.loc[
    ~customers["age"].between(18, 100),
    "age"
] = np.nan
```

```python
##Agent Cleaning
agents = agents.drop_duplicates()

agents = agents.drop_duplicates(
    subset="agent_id",
    keep="first"
)
```

```python
agents["agent_name"] = (
    agents["agent_name"]
    .astype("string")
    .str.strip()
)

agents["city"] = (
    agents["city"]
    .astype("string")
    .str.strip()
    .str.title()
)

agents["region"] = (
    agents["region"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Date
agents["join_date"] = pd.to_datetime(
    agents["join_date"],
    errors="coerce"
)
```

```python
##Accounts Cleaning
accounts = accounts.drop_duplicates()

accounts = accounts.drop_duplicates(
    subset="account_id",
    keep="first"
)
```

```python
accounts["account_type"] = (
    accounts["account_type"]
    .astype("string")
    .str.strip()
    .str.title()
)

accounts["status"] = (
    accounts["status"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Date
accounts["opening_date"] = pd.to_datetime(
    accounts["opening_date"],
    errors="coerce"
)
```

```python
##Check invalid Customer
invalid_accounts = accounts[
    ~accounts["customer_id"].isin(
        customers["customer_id"]
    )
]

print(invalid_accounts)
```

```python
##Then Remove
accounts = accounts[
    accounts["customer_id"].isin(
        customers["customer_id"]
    )
]
```

```python
##Transaction Types Cleaning
transaction_types = transaction_types.drop_duplicates()

transaction_types = transaction_types.drop_duplicates(
    subset="type_id",
    keep="first"
)
```

```python
transaction_types["transaction_type"] = (
    transaction_types["transaction_type"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Transactions Cleaning
transactions = transactions.drop_duplicates()
```

```python
transactions = transactions.drop_duplicates(
    subset="transaction_id",
    keep="first"
)
```

```python
## IDs
id_columns = [
    "transaction_id",
    "customer_id",
    "agent_id",
    "type_id"
]

for col in id_columns:
    transactions[col] = (
        transactions[col]
        .astype(str)
        .str.strip()
    )
```

```python
##Amount
transactions["amount"] = pd.to_numeric(
    transactions["amount"],
    errors="coerce"
)
```

```python
##Invalid Amount
transactions.loc[
    transactions["amount"] <= 0,
    "amount"
] = np.nan
```

```python
##Fee
transactions["fee"] = pd.to_numeric(
    transactions["fee"],
    errors="coerce"
)
```

```python
##Negative Fee
transactions.loc[
    transactions["fee"] < 0,
    "fee"
] = np.nan
```

```python
## Date & Time
```

```python
transactions["transaction_date"] = pd.to_datetime(
    transactions["transaction_date"],
    errors="coerce"
)
```

```python
##Time
transactions["transaction_time"] = pd.to_datetime(
    transactions["transaction_time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time
```

```python
##Status
transactions["status"] = (
    transactions["status"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Channel
transactions["channel"] = (
    transactions["channel"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

```python
##Foreign Key Validation
```

```python
##Customer validation
transactions[
    ~transactions["customer_id"].isin(
        customers["customer_id"]
    )
]
```

```python
##Agent validation
transactions[
    ~transactions["agent_id"].isin(
        agents["agent_id"]
    )
]
```

```python
##Transaction type validation
transactions[
    ~transactions["type_id"].isin(
        transaction_types["type_id"]
    )
]
```

```python
##valid transaction
transactions = transactions[
    transactions["customer_id"].isin(
        customers["customer_id"]
    )
    &
    transactions["agent_id"].isin(
        agents["agent_id"]
    )
    &
    transactions["type_id"].isin(
        transaction_types["type_id"]
    )
]
```

```python
##Derived Columns
```

```python
##Month
transactions["transaction_month"] = (
    transactions["transaction_date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
##Year
transactions["transaction_year"] = (
    transactions["transaction_date"].dt.year
)
```

```python
##Month Number
transactions["transaction_month_num"] = (
    transactions["transaction_date"].dt.month
)
```

```python
##Net Amount
transactions["net_amount"] = (
    transactions["amount"]
    - transactions["fee"].fillna(0)
)
```
